```python
import pandas as pd
import yfinance as yf


headlines= pd.read_csv("sp500_headlines_2008_2024.csv")
headlines['Date']= pd.to_datetime(headlines['Date'])

combined = headlines.groupby('Date').agg(
    combined_headline=('Title', lambda x: ' . '.join(x)),
    headline_count=('Title', 'count'),
    close_price=('CP', 'first')
).reset_index()

prices = pd.read_csv("sp500_prices.csv", header=0, skiprows=[1,2])
prices = prices.rename(columns={'Price': 'Date'})

prices['Date'] = pd.to_datetime(prices['Date'])
prices = prices.rename(columns={
    'Open': 'open_price',
    'High': 'high_price',
    'Low': 'low_price',
    'Close': 'close_price_check',
    'Volume': 'volume'
})

merged= combined.merge(
    prices[['Date', 'open_price', 'high_price', 'low_price', 'volume']],
    on='Date',
    how='inner'
)
merged =merged.sort_values('Date').reset_index(drop=True)
print(f"Final shape: {merged.shape}")
print(merged.head())
merged.to_csv("sp500_headlines_with_prices.csv", index=False)
```
